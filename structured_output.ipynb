## In this notebook we have tried to extract the output in a JSON format only by passing the response format in the agent's brain. So that we do not need to adjust the prompt for getting an accurate output format.

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.messages import TextMessage

from autogen_ext.models.openai import OpenAIChatCompletionClient

from autogen_core.models import ModelFamily

from autogen_core import CancellationToken

from pydantic import BaseModel

from dotenv import load_dotenv
import os

### Loading the api key--------------------------------------
load_dotenv()
api_key = os.getenv('GROQ_API_KEY')

## For structured output creating the pydantic class-----------
class PlanetInfo(BaseModel):
    name: str
    color: str
    distance_miles: int


## Initializing model client----------------------------------
custom_model_client = OpenAIChatCompletionClient(
    api_key=api_key,
    base_url= "https://api.groq.com/openai/v1",
    model="openai/gpt-oss-120b",
    # model= "llama-3.1-8b-instant",
    response_format=PlanetInfo,                 ####------- This will give us the structured output format
    model_info={
                    "vision": True,
                    "function_calling": True,
                    "json_output": False,
                    "family": ModelFamily.R1,
                    "structured_output": True,
                },
)
```

```python
agent = AssistantAgent(
    model_client=custom_model_client,
    name="demo_agent",
    system_message="A helpful agent",
    description="An agent who gives structured output"
)
```

```python
async def get_structured_output():
    task = TextMessage(content="Tell me about the Jupyter planet", source='user')
    response = await agent.run(task=task)
    final_response = response.messages[-1].content
    return final_response
```

```python
await get_structured_output()
```
