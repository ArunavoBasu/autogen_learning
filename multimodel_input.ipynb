### Autogen also has the capability to take multimodal input such as text as well as images and can give us a structured output

### But the model which we are using in the agent that must have the ability to take the image as the input. Such as "gpt-4o"

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_core.models import ModelFamily
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_agentchat.messages import UserMessage, TextMessage, MultiModalMessage
from autogen_core import Image as AGImage

from PIL import Image
from io import BytesIO
import requests
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv('GROQ_API_KEY')


## Initializing model client----------------------------------
custom_model_client = OpenAIChatCompletionClient(
    api_key=api_key,
    base_url= "https://api.groq.com/openai/v1",
    model="openai/gpt-oss-120b",
    # model= "llama-3.1-8b-instant",
    model_info={
                    "vision": True,
                    "function_calling": True,
                    "json_output": False,
                    "family": ModelFamily.R1,
                    "structured_output": True,
                },
)

agent = AssistantAgent(
    model_client=custom_model_client,
    name="demo_agent",
    system_message="use tools to solve tasks",
    description="An agent who is able to take multimodal input and gives response"
)
```

### MultiModal image as input

```python
async def test_multimodal():
    response = requests.get("https://picsum.photos/id/237/200/300")
    pil_image = Image.open(BytesIO(response.content))
    ag_image = AGImage(pil_image)

    multi_modal_msg = MultiModalMessage(
        content=['what is in the image?', ag_image],
        source='user'
    )

    result = await agent.run(task=multi_modal_msg)
    print(result.messages[-1].content)
```

```python
await test_multimodal()
```
